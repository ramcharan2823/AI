# **Lemmatization**

**WordNet Lemmatizer**


```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
## Q&A, Chatbots, Text Summerization
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS - Noun - n
Verb - v
adjective - a
adverb - r
'''
lemmatizer.lemmatize("going",pos = 'v')
```

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","goes","going","gone"]
```

```python
for word in words:
  print(word+"--->"+lemmatizer.lemmatize(word,pos='v'))
```
