**Named Entity Recognization**

```python
sentence = "The Eiffel Tower is made almost entirely of puddle iron (a form of wrought iron). It contains roughly 7,300 metric tons of iron, with a total structure weight of about 10,100 metric tons. Because iron undergoes thermal expansion, the tower can shrink or grow by several inches depending on the weather."
```

```python
import nltk
# nltk.download('punkt_tab')
# nltk.download('averaged_perceptron_tagger_eng')
# nltk.download('maxent_ne_chunker_tab')
# nltk.download('words')
words = nltk.word_tokenize(sentence)
```

```python
words
```

```python
tag_elements = nltk.pos_tag(words)
```

```python
# nltk.ne_chunk(tag_elements).draw() ## it will works in jupyter notebook
ner_tree = nltk.ne_chunk(tag_elements)
print(ner_tree)
```
